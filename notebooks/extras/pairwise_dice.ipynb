```python
# ── R18 · Pairwise agreement (Dice) between attribution methods ─────────────
#   For three methods, each pairwise Dice follows from the agreement scores:
#   a_m = mean Dice of method m with the other two  ->  D_xy = a_x + a_y - a_z
import glob, pandas as pd

path = (glob.glob("/kaggle/working/**/test_image_level.csv", recursive=True)
        + glob.glob("/kaggle/input/**/test_image_level.csv", recursive=True))[0]
print("Using:", path)
I = pd.read_csv(path)

I["D_gc_ig"] = I.a_gradcam + I.a_ig - I.a_smoothgrad
I["D_gc_sg"] = I.a_gradcam + I.a_smoothgrad - I.a_ig
I["D_ig_sg"] = I.a_ig + I.a_smoothgrad - I.a_gradcam

table = I.groupby("backbone_key")[["D_gc_ig", "D_gc_sg", "D_ig_sg"]].mean().round(3)
table.columns = ["Grad-CAM++ vs IG", "Grad-CAM++ vs SmoothGrad", "IG vs SmoothGrad"]
print(table)
table.to_csv("/kaggle/working/pairwise_dice.csv")
```
