```python
import glob
import pandas as pd
from statsmodels.stats.multitest import multipletests

BASE = "/kaggle/input/notebooks/tamannajarintima/benchmarking"
hits = glob.glob(f"{BASE}/**/mcnemar.csv", recursive=True)
print("Found:", hits)

df = pd.read_csv(hits[0])
df["p_holm"] = multipletests(df["p"], method="holm")[1]
df["significant_holm"] = df["p_holm"] < 0.05
df.to_csv("/kaggle/working/mcnemar_holm.csv", index=False)
print(df[["a", "b", "n10", "n01", "p", "p_holm", "significant_holm"]])
```
